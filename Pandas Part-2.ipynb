    Pandas Handling Missing Values
    

In Pandas, missing values, often represented as NaN (Not a Number), can cause problems during data processing and analysis. These gaps in data can lead to incorrect analysis and misleading conclusions.

Pandas provides a host of functions like dropna(), fillna() and combine_first() to handle missing values.

```python
import pandas as pd

# create dataframe with missing values
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, 5],
    'D': [1, 2, 3, 4, 5]
}
df = pd.DataFrame(data)

print(df)
```

```python
df.fillna(value=0, inplace=True)
print(df)
```

```python
df.dropna(inplace=True)#inplace=True specifies that changes are to be made in the original DataFrame itself.
print(df)
```

# Replace Missing Values With Mean, Median and Mode

```python
import pandas as pd
import numpy as np

# create a dataframe with missing values
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, 5],
    'D': [1, 2, 3, 4, 5]
}
df = pd.DataFrame(data)

# replace missing values with mean
df['A'].fillna(value=df['A'].mean(), inplace=True)

# replace missing values with median
df['B'].fillna(value=df['B'].median(), inplace=True)

# replace missing values with mode
df['C'].fillna(value=df['C'].mode()[0], inplace=True)

print(df)
```

```python
import pandas as pd

data = {
    "Product": [
        "Laptop", "Phone", "Laptop", "Headphones",
        "Phone", "Keyboard", "Headphones", "Laptop"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics", "Accessories",
        "Electronics", "Accessories", "Accessories", "Electronics"
    ],
    "Region": [
        "North", "South", "North", "East",
        "South", "East", "North", "South"
    ],
    "Quantity": [2, 5, 3, 8, 4, 10, 6, 1],
    "Unit_Price": [60000, 25000, 60000, 3000, 25000, 2000, 3000, 60000],
    "Rating": [4.5, 4.2, 4.7, 4.0, 4.4, 4.1, 4.3, 4.8]
}

df = pd.DataFrame(data)

print(df)
```

```python
import pandas as pd

# Sample dataset
data = {
    'Department': [ 'Finance', 'Finance', 'HR'],'Job_Role': [ 'Analyst', 'Analyst', 'Manager'],'Salary': [75000, 50000, 85000],'Years_Experience': [10, 5, 7]
}

df = pd.DataFrame(data)

# Group by Department and Job_Role, then calculate the sum of Salary
grouped = df.groupby(['Department', 'Job_Role']).agg({
    'Salary': 'sum',
    'Years_Experience': 'mean'
}).reset_index()

# Filter: Only include groups where the sum of Salary is greater than 150000
filtered = grouped[grouped['Salary'] > 100000]

print(filtered)
```

```python

import pandas as pd

# Create a DataFrame
var = pd.DataFrame({
    "days": [1, 2, 3, 4, 5, 6],
    "eng": [10, 12, 14, 15, 16, 12],
    "maths": [17, 18, 19, 13, 14, 16]
})

# Display the original DataFrame
print("Original DataFrame:")
print(var)

# Convert wide format to long format using melt()
melted_var = pd.melt(
    var,
    id_vars=["days"],
    var_name="subject",
    value_name="marks"
)

# Display the melted DataFrame
print("\nMelted DataFrame:")
print(melted_var)
```

In Python, melt() is a Pandas function that converts a DataFrame from wide format to long format.
In simple words, it converts multiple columns into two columns: one containing the column names and another containing their values.

```python
import pandas as pd
import numpy as np

# Sample Data
data = {
    'Region': ['North', 'North', 'South', 'South', 'East', 'East'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 250, 300, 400],
    'Quantity': [10, 12, 15, 18, 20, 25]
}
df = pd.DataFrame(data)

# 1. Single aggregation (Sum) across rows and columns
pivot_sum = df.pivot_table(index='Region', columns='Product', values='Sales', aggfunc='sum')

# 2. Multiple aggregations on a single value column
pivot_multi = df.pivot_table(index='Region', values='Sales', aggfunc=['mean', 'sum'])

# 3. Different aggregations mapped to different columns
pivot_dict = df.pivot_table(
    index='Region', 
    values=['Sales', 'Quantity'], 
    aggfunc={'Sales': 'sum', 'Quantity': 'mean'}
)
```
